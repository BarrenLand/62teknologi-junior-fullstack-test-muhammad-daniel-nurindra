```python

import cv2
import numpy as np

video_path = "(insert path here)"

cap = cv2.VideoCapture(video_path)
fps = cap.get(cv2.CAP_PROP_FPS)
frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
duration = frame_count / fps

# We'll sample fewer frames for speed
sample_rate = int(fps // 5) if fps > 5 else 1  # ~5 fps
brightness = []

frame_index = 0
while True:
    ret, frame = cap.read()
    if not ret:
        break
    if frame_index % sample_rate == 0:
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        brightness.append(np.mean(gray))
    frame_index += 1

cap.release()

brightness = np.array(brightness)
brightness = (brightness - np.min(brightness)) / (np.max(brightness) - np.min(brightness))

# Detect flashes by threshold crossings
threshold = 0.6
is_on = brightness > threshold

# Count rising edges (off -> on transitions)
flashes = np.count_nonzero((is_on[1:] == True) & (is_on[:-1] == False))

# Calculate flashes per minute
observed_duration = len(brightness) / (fps / sample_rate)
fpm = flashes / observed_duration * 60
fpm
```
